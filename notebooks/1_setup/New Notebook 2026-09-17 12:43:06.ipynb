```python
# creating a catalog names 'ecommerce'

spark.sql("CREATE CATALOG IF NOT EXISTS ecommerce")
```

```python
# use the ecmmerce catalog

spark.sql("USE CATALOG ecommerce")
```

Now we will create three schemas for bronze, silver, and gold layers maintaining Medallion architecture.

```python
%sql

CREATE SCHEMA IF NOT EXISTS ecommerce.bronze;
CREATE SCHEMA IF NOT EXISTS ecommerce.silver;
CREATE SCHEMA IF NOT EXISTS ecommerce.gold;
```

```python
%sql
---- to see the schemas in the Catalog ecommerce we can use SHOW DATABASES function in sql

SHOW DATABASES FROM ecommerce
```
